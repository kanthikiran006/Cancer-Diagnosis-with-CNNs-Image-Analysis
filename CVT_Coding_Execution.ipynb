COMPUTER VISION TECHNIQUES – UNIT-1 LAB

Step 1: Importing libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Step 2: Uploading and read image

```python
from google.colab import files
uploaded = files.upload()

img = cv2.imread(list(uploaded.keys())[0])
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Input Image")
plt.axis("off")
```

Step 3: Classical filtering operations

3.1 Mean filter

```python
mean = cv2.blur(gray, (5,5))

plt.imshow(mean, cmap="gray")
plt.title("Mean Filter")
plt.axis("off")
```

3.2 Gaussian filter

```python
gaussian = cv2.GaussianBlur(gray, (5,5), 0)

plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian Filter")
plt.axis("off")
```

3.3 Median filter

```python
median = cv2.medianBlur(gray, 5)

plt.imshow(median, cmap="gray")
plt.title("Median Filter")
plt.axis("off")
```

3.4 Bilateral filter

```python
bilateral = cv2.bilateralFilter(gray, 9, 75, 75)

plt.imshow(bilateral, cmap="gray")
plt.title("Bilateral Filter")
plt.axis("off")
```

Step 4: Thresholding techniques

4.1 Binary thresholding

```python
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(binary, cmap="gray")
plt.title("Binary Thresholding")
plt.axis("off")
```

4.2 Adaptive thresholding

```python
adaptive = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11, 2)

plt.imshow(adaptive, cmap="gray")
plt.title("Adaptive Thresholding")
plt.axis("off")
```

4.3 Otsu thresholding

```python
_, otsu = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.imshow(otsu, cmap="gray")
plt.title("Otsu Thresholding")
plt.axis("off")
```

Step 5: Edge detection techniques

5.1 Sobel edge detection

```python
sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sx, sy)

plt.imshow(sobel, cmap="gray")
plt.title("Sobel Edge Detection")
plt.axis("off")
```

5.2 Canny edge detection

```python
canny = cv2.Canny(gray, 100, 200)

plt.imshow(canny, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")
```

Step 6: Corner & interest point detection

6.1 Harris corner detection

```python
harris_img = img.copy()
gray_float = np.float32(gray)

harris = cv2.cornerHarris(gray_float, 2, 3, 0.04)
harris_img[harris > 0.01 * harris.max()] = [0,0,255]

plt.imshow(cv2.cvtColor(harris_img, cv2.COLOR_BGR2RGB))
plt.title("Harris Corner Detection")
plt.axis("off")
```

6.2 Shi-Tomasi corner detection

```python
corners = cv2.goodFeaturesToTrack(gray, 100, 0.01, 10)
corners = np.int32(corners)

shi_img = img.copy()
for c in corners:
    x, y = c.ravel()
    cv2.circle(shi_img, (x,y), 3, (0,255,0), -1)

plt.imshow(cv2.cvtColor(shi_img, cv2.COLOR_BGR2RGB))
plt.title("Shi-Tomasi Corner Detection")
plt.axis("off")
```

Final observation

Filtering smooths the image and reduces noise.

Thresholding separates foreground and background.

Edge detection highlights object boundaries.

Corner detection identifies interest points.
